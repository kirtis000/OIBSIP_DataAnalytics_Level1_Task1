## Monthly Sales Trend

This analysis examines how total sales changed over time.
It helps identify months with higher or lower sales activity.

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()

monthly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Correlation Analysis

The strongest correlation is between Price per Unit and Total Amount,
with a correlation coefficient of approximately 0.85. This is a strong
positive relationship and is expected because total transaction amount
depends directly on the price per unit and quantity.

Quantity and Total Amount have a moderate positive correlation of
approximately 0.37, indicating that larger quantities tend to be
associated with higher transaction amounts.

Age has very weak correlations with Quantity, Price per Unit, and
Total Amount, with all coefficients close to zero. Therefore, age does
not show a meaningful linear relationship with transaction value in
this dataset.

### Additional Visualization: Average Transaction Amount by Age Group

The Under 18 group has the highest average transaction amount,
while the 56+ group has the lowest average transaction amount.

The average transaction amount generally decreases across the age
groups in this dataset. However, the Under 18 group contains only

because it is based on a relatively small number of observations.

# Business Insights and Recommendations

## Key Findings

1. May 2023 recorded the highest monthly sales at 53,150,
   while September 2023 recorded the lowest complete-month sales
   at 23,620.

2. Q4 2023 generated the highest quarterly sales at 126,190,
   followed by Q2 with 123,735.

3. Electronics generated the highest category revenue at 156,905,
   followed closely by Clothing at 155,580.

4. Female transactions represented 51% of the dataset, while
   male transactions represented 49%, showing a relatively
   balanced gender distribution.

5. The 46–55 age group had the highest number of transactions,
   with 229 transactions.

6. Price per Unit had a strong positive correlation with Total
   Amount (approximately 0.85).

7. Quantity had a moderate positive correlation with Total Amount
   (approximately 0.37).

8. Age had very weak linear correlation with Total Amount,
   approximately -0.06.

9. The Under 18 group showed the highest average transaction
   amount, but this result should be interpreted cautiously
   because the group contains only 21 transactions.

## Business Recommendations

### 1. Focus on high-performing sales periods

The business should examine the factors behind the strong sales
performance in May and Q4 2023. Successful promotions, product
availability, and seasonal demand during these periods can be
studied and potentially replicated in weaker months.

### 2. Maintain strong focus on Electronics and Clothing

Electronics generated the highest revenue, while Clothing was
very close behind. The business could maintain sufficient inventory
for these categories and use promotions or product bundles to
increase sales while monitoring their margins.

### 3. Use age-based analysis for targeted campaigns

The 46–55 age group generated the largest number of transactions,
while transaction value varied across age groups. Marketing campaigns
can therefore be evaluated by age segment rather than treating all
customers identically.

Additional data should be collected before making strong conclusions
about the spending behavior of smaller groups such as customers
under 18.
